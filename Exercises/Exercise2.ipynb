```python
#Logical Operators

# AND
a = 10
b = 20
c = 30

print(a<b and a<c)
print(a<b and a>c)
print(a>b and a>c)
print(a>b and c>b) 
print()

# OR

print(a<b or a<c)
print(a<b or a>c)
print(a>b or a<c)
print(a>b or c<b)
print()

# Not

print(a>b)
print(not(a>b))
```

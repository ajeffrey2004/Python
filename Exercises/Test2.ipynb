```python
#2
Values = (10, 20, 30, 40, 50)
print(Values)
Values_list = list(Values)
print(Values_list)
Values_list[2] = 35
print(Values_list)
Values_list.append(60)
print(Values_list)
Values_list.pop(0)
print(Values_list)
Values = tuple(Values_list)
print(Values)
```

```python
#4
morning_batch = {"Arun", "Bala", "Charan", "Divya"}
evening_batch = {"Bala", "Divya", "Ezhil", "Farah"}

print(morning_batch.union(evening_batch))
print(morning_batch.intersection(evening_batch))
print(morning_batch.difference(evening_batch))
print(evening_batch.difference(morning_batch))
```

```python
#6
available_products = {"Laptop", "Mouse", "Keyboard"}
new_products = {"Monitor", "Mouse", "Printer"}

available_products.add("Webcam")
print(available_products)
available_products.remove("Keyboard")
print(available_products)
Products = available_products.union(new_products)
print(Products)
common_products = available_products.intersection(new_products)
print(common_products)
unique_available_products = available_products.difference(new_products)
print(unique_available_products)
```

```python
#7
marks = {"Arun": 85, "Bala": 92, "Charan": 78, "Divya": 88}

print("Mark of Bala:", marks["Bala"])
print("Mark of Charan:", marks.get("Charan"))
print("Student names:", list(marks.keys()))
print("Marks:", list(marks.values()))
print("Student names and marks:", list(marks.items()))
```

```python
#8
department_staff = {
    "HR": ["Ravi", "Sita"],
    "IT": ["Varun", "Ananya"],
    "Sales": ["Kiran", "Meena"]
}

print("Complete dictionary:", department_staff)
print("Employees in IT department:", department_staff["IT"])
print("First employee in HR department:", department_staff["HR"][0])
print("Last employee in Sales department:", department_staff["Sales"][-1])
department_staff["HR"].append("Madhu")
print("Complete dictionary:", department_staff)
```

```python
#9
cart = {"Pen": 10, "Notebook": 50, "Bag": 500}

cart["Pencil"] = 5
print(cart)
cart.update({"Notebook": 60})
print(cart)
print(cart.get("Bag"))
print(cart.pop("Pen"))
print(cart)
cart.clear()
print(cart)
```

```python
#10
subjects = ["Tamil", "English", "Math", "Science"]
marks = [90, 85, 95, 88]

dictionary = dict(zip(subjects, marks))
print(dictionary)
dictionary.setdefault("Social", 80)
print(dictionary)
dictionary.setdefault("Math", 100)
print(dictionary) #Existing Math mark doesn't changes.
```
